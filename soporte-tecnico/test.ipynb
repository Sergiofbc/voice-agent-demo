```python
from dotenv import load_dotenv
from elevenlabs.client import ElevenLabs
import os

load_dotenv()

client = ElevenLabs(api_key=os.getenv("ELEVENLABS_API_KEY_FULL"))

voices = client.voices.get_all()

for voice in voices.voices:
    langs = getattr(voice, "verified_languages", [])

    if any(getattr(lang, "language", "") == "es" for lang in langs):
        print("=" * 80)
        print("Nombre:", voice.name)
        print("ID:", voice.voice_id)
        print("Categoría:", voice.category)
```
